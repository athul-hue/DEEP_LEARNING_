```python
import numpy as np
import matplotlib.pyplot as plt
```

X represents the input and Y represents the actual output.

```python
# Data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 6, 8, 10])

```

w is the weight.

b is the bias.

We don't know the correct values initially, so we start with zero.

```python
# Initial values
w = 0
b = 0

learning_rate = 0.01
iterations = 30

losses = []

# Gradient Descent
for i in range(iterations):

    # Prediction
    Y_pred = w * X + b

    # Calculate error
    error = Y_pred - Y

    # Calculate loss
    loss = np.mean(error ** 2)
    losses.append(loss)

    # Calculate gradients
    dw = np.mean(2 * X * error)
    db = np.mean(2 * error)

    # Update weight and bias
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # Print every 10 iterations
    if i % 10 == 0:
        print("Iteration:", i, "Loss:", round(loss, 4))

print("\nFinal Weight:", round(w, 2))
print("Final Bias:", round(b, 2))

```

```python
# Show loss reduction
plt.plot(losses)
plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Gradient Descent")
plt.show()


```
